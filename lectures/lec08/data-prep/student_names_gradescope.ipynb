```python
import pandas as pd
import numpy as np
```

Download the roster from Gradescope. **Don't** commit the roster to GitHub – let's avoid the possibility of accidentally releasing it to students.

```python
roster = pd.read_csv('gradescope_enrolled.csv')
roster
```

```python
# drop first two rows because First Name is NaN and -
# handle this manually each quarter
roster = roster.take(np.arange(2, roster.shape[0]))
```

```python
roster['Section'].value_counts()
```

```python
# The lecture number is in positions 10, 11, and 12
# Example: 'DSC-010 - 004-001-DI' ==> 004
roster['Lecture'] = roster['Section'].str[10:13]
roster
```

```python
# For anonymity, use random last names
last_names = np.asarray(list(map(lambda x: ''.join(x).title(), 
                                 np.random.choice(list('abcdefghijklmnopqrstuvwxyz'), 
                                size=(roster.shape[0], 6)))))
last_names
```

```python
roster['Student'] = (roster.get('First Name') + ' ' +  last_names)
roster
```

```python
roster = roster.get(['Student', 'Lecture', 'First Name'])
roster.columns=['name', 'section', 'first']
roster
```

```python
roster = roster.sample(frac=1).reset_index(drop=True) # So it's not alphabetically sorted
roster
```

```python
roster.to_csv('../data/roster-anon.csv', index=False)
```

```python
def first_name(full_name):
    return full_name.split(' ')[0]
```

```python
with_first = roster.assign(first=roster['name'].apply(first_name))
```

```python
first_counts = with_first.groupby('first').count().sort_values('name', ascending=False)
```

```python
def how_many_others_in_section(name):
    row = with_first[roster['name'] == name]
    first = row.get('first').iloc[0]
    section = row.get('section').iloc[0]
    return with_first[(with_first.get('section') == section) & (with_first.get('first') == first)].shape[0]
```

```python
with_others = with_first.assign(others=with_first.get('name').apply(how_many_others_in_section))
with_others
```

```python
with_others[with_others.get('others') > 1].sort_values('others', ascending=False)
```
