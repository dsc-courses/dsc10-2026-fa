We'll define the age of a person as 2026 (the current year) minus the year in which the person was born. This doesn't take into account people's birthdays, because we don't have that information. For example, if a female Kevin was born in 1985, they will be counted as 2026 - 1985 = 41 years old. Therefore the **total age** of all the female Kevins is given below.

````python
total_age = ((2026 - female_kevins.get('Year')) * female_kevins.get('Count')).sum()
total_age
````

To find the average age, we need to know how many female Kevins there are. The **total count** of female Kevins is given below.

````python
total_count = female_kevins.get('Count').sum()
total_count
````

Therefore the **average age** of female Kevins is given below.

````python
average_age = total_age / total_count
average_age
````
